### **Small CNN**

```python
!python {BASE}/baseline_cnn.py \
    --train_dir {BASE}/train \
    --test_dir  {BASE}/test \
    --folds     {BASE}/baseline_rf_out_v2/fold_assignments.csv \
    --out_dir   {BASE}/baseline_out_cnn \
    --epochs    40 \
    --patience  8 \
    --batch_size 64 \
    --num_workers 2
```

**Probability ensemble: LGBM-v2 + Small CNN (v1)**

It tries 21 blend ratios (w * LGBM + (1-w) * CNN) ranging from pure-CNN to pure-LGBM, scoring each on the OOF training predictions. Whichever blend gives the highest macro F1 wins. This step is purely on training data — no test info touched, so it's an honest estimate of how the ensemble would perform.
